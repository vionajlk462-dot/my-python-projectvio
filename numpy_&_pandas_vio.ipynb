# **NUMPY**

```python
import numpy as np

#menggunakan kurung siku tunggal untuk memvisualisasikan secara vektor
vektor1= np.array([2,4,6,8])
print(vektor1)
print(vektor1.shape)

#menggunakan kurung siku ganda untuk membuat matriks
matriks1= np.array([[2,3,4,],[8,12,14]])
print(matriks1)
print(matriks1.shape)

```

```python
import numpy as np
#mengubah ukuran matriks menggunakan fungsi reshape
matriks2= np.array([[2,3,4],[8,12,14]])
print('2,3',matriks2.shape)
matriksnew=matriks2.reshape(3,2)
print(matriksnew)

#membuat matriks 3 dimensi
#matriksnew2=matriks2.reshape(2,4,3)
#tidak bisa membuat 3 dimensi karena size yang ada terlalu kecil sehingga saya harus reshape ulang/menambah data di array

```

```python
import numpy as np
#membuat data dengan menggunakan fungsi np.arange
range1=np.arange(0, 8)
print(range1)

#membuat matriks mengggunakan fungsi np.arange
range2=np.arange(0,8).reshape(4,2)
print(range2)
```

```python
import numpy as np
#membuat matriks kelipatan menggunakan np.arange
kelipatan1=np.arange(8,80,8) # Membuat angka kelipatan 8, mulai dari 8 sampai sebelum 80
print(kelipatan1)

#membuat matriks kelipatan  menggunakan np.arange dan memakai fungsi reshape
kelipatan2=np.arange(2,60,4).reshape(5,3)
print(kelipatan2)
```

```python
import numpy as np
#Membuat slicing dan indexing
nampay1=([1,2,3,4,5])
print(nampay1[3]) #indexing(mengambil 1 elemen dalam array)
print(nampay1[2:]) #mulai dari posisi 2 hingga posisi terakhir
print(nampay1[-2:]) #mengambil 2 elemen terakhir dari kanan ke kiri
print(nampay1[2:4]) #slicing(mengambil n-rentang elemen dalam array)
```

```python
import numpy as np
urutan, jarak = np.linspace(0, 1, 5, retstep=True)
print("Urutan:", urutan)
print("Jarak antara elemen:", jarak)
```

# **PANDAS**

```python

#menjalankan pandas dan numpy
import pandas as pd
import numpy as np
#menampilkan array 1 dimensi
ds = pd.Series([2, 4, 6, 8, 10])
print(ds)
```

```python
data = pd.Series(['a','b','c'])
np1 = np.array(['x','y','z'])
data_np = pd.Series(np1)
print(data)
print(data_np)
#mengatur indeks pada numpy array data-np di atas
data_np = pd.Series(np1, index=[2,3,4])
print(data_np)
```

```python
ds1 = pd.Series([2, 4, 6, 8, 10])
ds2 = pd.Series([1, 3, 5, 7, 9])
#operasi jumlah
dsa = ds1 + ds2
dsb = ds1 - ds2
dsc = ds1 * ds2
dsd = ds1 / ds2
print(dsa)
print(dsb)
print(dsc)
print(dsd)
```

```python
#konversi pandas module Series ke python list
import pandas as pd
ds = pd.Series([2, 4, 6, 8, 10])
print("Pandas Series dan type")
print(ds)
print(type(ds))
print("Konver Pandas Series ke Python list")
print(ds.tolist())
print(type(ds.tolist()))
```

```python
#konversi a NumPy array to a Pandas series
import numpy as np
import pandas as pd

np_array = np.array([0, 20, 30, 40, 50])
print("NumPy array:")
print(np_array)
new_np = pd.Series(np_array)
print("konversi Pandas series:")
print(new_np)
```

```python
import pandas as pd
import numpy as np
#generate angka random
num_state = np.random.RandomState(100)
num_series = pd.Series(num_state.normal(10, 4, 20))
print(num_series)
print(num_state)
```

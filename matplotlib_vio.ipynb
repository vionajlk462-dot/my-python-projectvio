```python
import matplotlib.pyplot as plt

# Data
bulan = ['Jan', 'Feb', 'Mar', 'Apr', 'Mei']
Penjualan = [150, 200, 180, 250, 125]

# Ukuran grafik
plt.figure(figsize=(8, 4))

# Membuat grafik batang
plt.barh(bulan, Penjualan, color='blue')

# Judul dan label
plt.title('Penjualan per Bulan')
plt.xlabel('Bulan')
plt.ylabel('Jumlah')

# Menampilkan grafik
plt.show()

# Menambahkan nilai angka di ujung setiap batang
for i, nilai in enumerate(Penjualan):
    plt.text(i, nilai + 2, str(nilai),
             ha='center',
             va='center',
             fontsize=10,
             fontweight='bold')

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Data
bulan = ['Jan', 'Feb', 'Mar', 'Apr', 'Mei', 'Juni', 'Juli']
penjualan = [150, 200, 180, 250, 125, 220, 310]

# Ukuran gambar (diperlebar agar 7 bulan muat dengan rapi)
plt.figure(figsize=(8, 4))

# 1. Buat Grafik Batang
sns.barplot(x=bulan, y=penjualan, hue=bulan, palette='muted', legend=False)

# 2. Tambahkan Garis Tren (Menghubungkan puncak batang)
# range(len(bulan)) menghasilkan [0, 1, 2, 3, 4, 5, 6] sebagai posisi sumbu X
plt.plot(
    range(len(bulan)),  # Posisi X (tengah setiap batang)
    penjualan,          # Posisi Y (nilai penjualan)
    color='red',        # Warna garis
    marker='o',         # Tambahkan titik bulat di setiap persimpangan
    linestyle='-',      # Gaya garis (solid)
    linewidth=2,        # Ketebalan garis
    label='Garis Tren'  # Nama untuk legenda
)

# 3. Kustomisasi Tampilan
plt.title('Penjualan per Bulan dengan Garis Tren', fontsize=12, fontweight='bold')
plt.xlabel('Bulan')
plt.ylabel('Jumlah Penjualan')
plt.legend()  # Menampilkan kotak legenda

plt.tight_layout()
plt.show()
```

```python
import plotly.express as px
import pandas as pd

# Data (Plotly lebih suka data dalam bentuk DataFrame)
df = pd.DataFrame({
    'Bulan': ['Jan', 'Feb', 'Mar', 'Apr', 'Mei'],
    'Penjualan': [150, 200, 180, 250, 300]
})

# Membuat grafik dasar
fig = px.bar(
    df,
    x='Bulan',
    y='Penjualan',
    title='Penjualan per Bulan (Plotly)'
)

fig.update_layout(width=700, height=400)  # pixel

# Menampilkan grafik
fig.show()
```

```python
## menggunakan data set 1
# 1. Muat dataset
url = "https://raw.githubusercontent.com/Alam-A99/Pandas_Latih/10ebb3af000e930a44989075733f495d5c2f4be5/penjualan.csv"
df = pd.read_csv(url)
df

# 2. Tambahkan kolom total untuk memudahkan beberapa grafik
df['total'] = df[['jan', 'feb', 'mar', 'apr']].sum(axis=1)

# Atur style seaborn agar lebih rapi
sns.set_theme(style="whitegrid")

# Bar Chart
plt.figure(figsize=(8, 5)) # Ukuran kompak (lebar 8, tinggi 5 inci)

# Urutkan data dari total tertinggi
df_sorted = df.sort_values('total', ascending=False)

sns.barplot(data=df_sorted, x='nama', y='total', palette='viridis')
plt.title('Total Penjualan per Orang', fontsize=12)
plt.xlabel('Nama Karyawan')
plt.ylabel('Total Penjualan')
plt.xticks(rotation=45) # Miringkan label agar tidak tumpang tindih
plt.tight_layout()
plt.show()
```

```python
# Pie chart
plt.figure(figsize=(5, 5))  # Ukuran persegi agar pie chart tidak gepeng

# Hitung total penjualan per bulan
total_per_bulan = df[['jan', 'feb', 'mar', 'apr']].sum()

plt.pie(
    total_per_bulan,
    labels=total_per_bulan.index,
    autopct='%1.1f%%',
    startangle=140,
    colors=sns.color_palette("muted")
)

plt.title('Proporsi Penjualan per Bulan', fontsize=12)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import plotly.express as px

# Ubah format data menjadi 'long' (khusus mengambil kolom 'jan' dan 'mar')
df_melted = df.melt(
    id_vars='nama',
    value_vars=['jan', 'mar'],
    var_name='Bulan',
    value_name='Penjualan'
)

# Buat Line Chart (Hanya butuh 1 blok kode inti!)
fig = px.line(
    df_melted,
    x='nama',
    y='Penjualan',
    color='Bulan',
    symbol='Bulan',
    symbol_sequence=['circle', 'triangle-up'],
    color_discrete_sequence=['royalblue', 'crimson'],
    markers=True,
    title='Perbandingan Penjualan: Januari vs Maret',
    labels={
        'nama': 'Nama Karyawan',
        'Penjualan': 'Jumlah Penjualan'
    }
)

# Atur ukuran agar tidak full screen (Opsional)
fig.update_layout(width=800, height=500)

# Tampilkan grafik
fig.show()
```
